## ⚠️ 2. Data Leakage Check — Read This First

Before any analysis, identify columns that **directly encode the outcome**.
Using them as features would give your model information it could never have
at prediction time (i.e., before the booking resolves).

These must be dropped before any modeling.

```python
# reservation_status encodes the outcome directly:
# 'Canceled' means is_canceled=1, 'Check-Out' or 'No-Show' means is_canceled=0
print("reservation_status vs is_canceled:")
print(pd.crosstab(df["reservation_status"], df["is_canceled"]))
print()
print("=> reservation_status and reservation_status_date are LEAKY.")
print("   They must be dropped before any model training.")
print("   This is already handled in HotelBookingLoader._preprocess().")
```

## 3. Target Variable: `is_canceled`

```python
target_counts = df["is_canceled"].value_counts()
target_pct = df["is_canceled"].value_counts(normalize=True) * 100

print("Class distribution:")
for label, count in target_counts.items():
    name = "Not Canceled" if label == 0 else "Canceled"
    print(f"  {name} ({label}): {count:,} ({target_pct[label]:.1f}%)")

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Bar chart
axes[0].bar(["Not Canceled", "Canceled"], target_counts.values,
            color=["#4C72B0", "#DD8452"], edgecolor="white", linewidth=1.5)
axes[0].set_title("Class Distribution", fontweight="bold")
axes[0].set_ylabel("Count")
for i, v in enumerate(target_counts.values):
    axes[0].text(i, v + 200, f"{v:,}", ha="center", va="bottom", fontsize=10)

# Pie chart
axes[1].pie(target_pct.values, labels=["Not Canceled", "Canceled"],
            autopct="%1.1f%%", colors=["#4C72B0", "#DD8452"],
            startangle=90, wedgeprops=dict(edgecolor="white", linewidth=2))
axes[1].set_title("Class Balance", fontweight="bold")

plt.tight_layout()
plt.savefig("../docs/target_distribution.png", bbox_inches="tight")
plt.show()
```

## 4. Missing Values

```python
null_counts = df.isnull().sum()
null_pct = (df.isnull().mean() * 100).round(2)
missing_df = pd.DataFrame({"count": null_counts, "pct": null_pct})
missing_df = missing_df[missing_df["count"] > 0].sort_values("pct", ascending=False)

if missing_df.empty:
    print("No missing values found.")
else:
    print("Columns with missing values:")
    print(missing_df.to_string())

    fig, ax = plt.subplots(figsize=(8, max(3, len(missing_df) * 0.5)))
    ax.barh(missing_df.index, missing_df["pct"], color="#C44E52", edgecolor="white")
    ax.set_xlabel("% Missing")
    ax.set_title("Missing Values by Column", fontweight="bold")
    for i, (idx, row) in enumerate(missing_df.iterrows()):
        ax.text(row["pct"] + 0.1, i, f"{row['pct']:.1f}%", va="center", fontsize=9)
    plt.tight_layout()
    plt.savefig("../docs/missing_values.png", bbox_inches="tight")
    plt.show()
```

### 📝 Observation

*Write your observation here. Questions to answer:*
- *For `children`: what does a null mean? Probably 0 — impute with 0, not mean.*
- *For `country`: null could mean unknown origin. Treat as its own category ('UNK')?*
- *Is missingness random (MCAR), or does it correlate with the target (MAR/NMAR)?*

```python
# Check if missingness correlates with the target
# This matters: if missing values are more common in canceled bookings,
# the missingness itself is a feature.
for col in missing_df.index:
    null_mask = df[col].isnull()
    cancel_rate_missing = df.loc[null_mask, "is_canceled"].mean()
    cancel_rate_present = df.loc[~null_mask, "is_canceled"].mean()
    print(f"{col}: cancel rate when null={cancel_rate_missing:.3f}, "
          f"when present={cancel_rate_present:.3f}  "
          f"(diff={abs(cancel_rate_missing - cancel_rate_present):.3f})")
```

## 5. Numerical Features

```python
# Exclude the target and leaky columns
LEAKY = ["is_canceled", "reservation_status", "reservation_status_date"]
numerical_cols = [c for c in df.select_dtypes(include="number").columns
                  if c not in LEAKY]
print(f"{len(numerical_cols)} numerical features: {numerical_cols}")
```

```python
# Distribution plots for each numerical feature, split by target
# A good feature shows different distributions for canceled vs not canceled.

n_cols = 3
n_rows = (len(numerical_cols) + n_cols - 1) // n_cols
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, n_rows * 3.5))
axes = axes.flatten()

for i, col in enumerate(numerical_cols):
    ax = axes[i]
    for label, color in [(0, "#4C72B0"), (1, "#DD8452")]:
        subset = df.loc[df["is_canceled"] == label, col].dropna()
        # Clip extreme outliers for visualization only
        p99 = subset.quantile(0.99)
        subset = subset[subset <= p99]
        ax.hist(subset, bins=40, alpha=0.55, color=color,
                label=("Not Canceled" if label == 0 else "Canceled"),
                density=True)
    ax.set_title(col, fontweight="bold", fontsize=9)
    ax.set_xlabel("")
    ax.legend(fontsize=7)
    ax.tick_params(labelsize=7)

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle("Numerical Feature Distributions by Cancellation Status",
             fontsize=13, fontweight="bold", y=1.01)
plt.tight_layout()
plt.savefig("../docs/numerical_distributions.png", bbox_inches="tight")
plt.show()
```

```python
# Outlier summary using IQR method
print(f"{'Column':<35} {'Min':>10} {'Max':>12} {'IQR Outliers':>14} {'% Outliers':>11}")
print("-" * 85)

for col in numerical_cols:
    s = df[col].dropna()
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    outliers = ((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).sum()
    pct = outliers / len(s) * 100
    flag = " ⚠️" if pct > 5 else ""
    print(f"{col:<35} {s.min():>10.1f} {s.max():>12.1f} {outliers:>14,} {pct:>10.1f}%{flag}")
```

- *`adr` (average daily rate): are the extreme values data errors or legitimate luxury bookings?*

- *Which numerical features show the most separation between canceled and not canceled?*

## 6. Categorical Features

Using our `FrequencyMap` class from `src/utils/frequency_map.py` to profile each feature.

```python
categorical_cols = [c for c in df.select_dtypes(include="object").columns
                    if c not in LEAKY]
print(f"{len(categorical_cols)} categorical features:")

print(f"\n{'Column':<30} {'Unique':>7} {'Top Value':<25} {'Top %':>7}")
print("-" * 72)

for col in categorical_cols:
    fm = FrequencyMap(df[col].dropna().tolist())
    summary = fm.summary()
    print(f"{col:<30} {summary['unique']:>7} "
          f"{str(summary['top']):<25} {summary['top_proportion']*100:>6.1f}%")
```

```python
# Cancellation rate by category — the most useful signal for feature importance intuition
# Features where cancellation rate varies significantly across categories are predictive.

LOW_CARDINALITY = [c for c in categorical_cols
                   if df[c].nunique() <= 10]

n_cols = 2
n_rows = (len(LOW_CARDINALITY) + 1) // n_cols
fig, axes = plt.subplots(n_rows, n_cols, figsize=(14, n_rows * 4))
axes = axes.flatten()

for i, col in enumerate(LOW_CARDINALITY):
    ax = axes[i]
    cancel_rate = (df.groupby(col)["is_canceled"]
                   .mean()
                   .sort_values(ascending=True))
    colors = ["#C44E52" if v > 0.4 else "#4C72B0" if v < 0.25 else "#8172B2"
              for v in cancel_rate.values]
    cancel_rate.plot(kind="barh", ax=ax, color=colors, edgecolor="white")
    ax.set_title(f"Cancellation Rate by {col}", fontweight="bold", fontsize=9)
    ax.set_xlabel("Cancellation Rate")
    ax.axvline(x=df["is_canceled"].mean(), color="gray",
               linestyle="--", linewidth=1, label="Overall avg")
    ax.legend(fontsize=7)
    ax.tick_params(labelsize=8)
    ax.xaxis.set_major_formatter(mticker.PercentFormatter(xmax=1))

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle("Cancellation Rate by Categorical Feature",
             fontsize=13, fontweight="bold", y=1.01)
plt.tight_layout()
plt.savefig("../docs/categorical_cancel_rates.png", bbox_inches="tight")
plt.show()
```

```python
# High-cardinality features (country) — top 15 by volume, with cancel rates
HIGH_CARDINALITY = [c for c in categorical_cols if df[c].nunique() > 10]
print(f"High-cardinality columns (>10 unique values): {HIGH_CARDINALITY}")

for col in HIGH_CARDINALITY:
    print(f"\nTop 15 values for '{col}' by booking volume:")
    top_vals = df[col].value_counts().head(15).index
    subset = df[df[col].isin(top_vals)]
    rates = (subset.groupby(col)
             .agg(count=("is_canceled", "count"),
                  cancel_rate=("is_canceled", "mean"))
             .sort_values("count", ascending=False))
    rates["cancel_rate"] = rates["cancel_rate"].apply(lambda x: f"{x:.1%}")
    print(rates.to_string())
```

## 7. Temporal Patterns

```python
# Month ordering for plots
month_order = ["January","February","March","April","May","June",
               "July","August","September","October","November","December"]

monthly = (df.groupby("arrival_date_month")["is_canceled"]
           .agg(["count", "mean"])
           .reindex(month_order)
           .rename(columns={"count": "bookings", "mean": "cancel_rate"}))

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 7), sharex=True)

ax1.bar(monthly.index, monthly["bookings"], color="#4C72B0", edgecolor="white")
ax1.set_title("Bookings by Arrival Month", fontweight="bold")
ax1.set_ylabel("Number of Bookings")
ax1.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{int(x):,}"))

ax2.plot(monthly.index, monthly["cancel_rate"], marker="o",
         color="#DD8452", linewidth=2, markersize=6)
ax2.axhline(y=df["is_canceled"].mean(), color="gray",
            linestyle="--", linewidth=1, label="Overall avg")
ax2.set_title("Cancellation Rate by Arrival Month", fontweight="bold")
ax2.set_ylabel("Cancellation Rate")
ax2.yaxis.set_major_formatter(mticker.PercentFormatter(xmax=1))
ax2.legend()
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.savefig("../docs/temporal_patterns.png", bbox_inches="tight")
plt.show()
```

### 📝 Observation

*Are cancellation rates seasonal? Do peak summer months have higher or lower rates?*
*What does this imply for the `arrival_date_month` feature in Week 2?*

## 8. Correlation with Target

```python
# Point-biserial correlation: numerical feature vs binary target
# (equivalent to Pearson when one variable is binary)
corr_with_target = (
    df[numerical_cols + ["is_canceled"]]
    .corr()["is_canceled"]
    .drop("is_canceled")
    .sort_values(key=abs, ascending=False)
)

fig, ax = plt.subplots(figsize=(8, max(4, len(corr_with_target) * 0.4)))
colors = ["#C44E52" if v > 0 else "#4C72B0" for v in corr_with_target.values]
corr_with_target.plot(kind="barh", ax=ax, color=colors, edgecolor="white")
ax.set_title("Correlation of Numerical Features with is_canceled",
             fontweight="bold")
ax.set_xlabel("Pearson Correlation")
ax.axvline(x=0, color="black", linewidth=0.8)
plt.tight_layout()
plt.savefig("../docs/feature_correlations.png", bbox_inches="tight")
plt.show()

print("\nTop positive correlations (→ more likely to cancel):")
print(corr_with_target[corr_with_target > 0].head(5).to_string())
print("\nTop negative correlations (→ less likely to cancel):")
print(corr_with_target[corr_with_target < 0].head(5).to_string())
```

### 📝 Observation

*Which numerical features correlate most strongly with cancellation?*
*Are any correlations surprising? Are any surprisingly weak?*
*Note: correlation measures linear relationships only. A feature with low linear*
*correlation may still be highly predictive via a tree-based model.*
